```python
import neurokit2 as nk
import matplotlib.pyplot as plt

ecg = nk.ecg_simulate(duration=10, sampling_rate=300, heart_rate=70, noise=0.05, random_state=0)
signals, info = nk.ecg_process(ecg, sampling_rate=300)
nk.ecg_plot(signals, info)
fig = plt.gcf()                 
fig.set_size_inches(15, 8)      
fig.tight_layout()  
plt.show()

hrv = nk.hrv_time(info["ECG_R_Peaks"], sampling_rate=300)
print(hrv[["HRV_MeanNN", "HRV_SDNN", "HRV_RMSSD", "HRV_pNN50"]])
```

```python
ecg_noisy = nk.ecg_simulate(duration=10, sampling_rate=300, heart_rate=70, noise=0.5, random_state=0)
signals_n, info_n = nk.ecg_process(ecg_noisy, sampling_rate=300)
nk.ecg_plot(signals_n, info_n)
fig = plt.gcf()                 
fig.set_size_inches(15, 8)      
fig.tight_layout()
plt.show()

hrv_n = nk.hrv_time(info_n["ECG_R_Peaks"], sampling_rate=300)
print(hrv_n[["HRV_MeanNN", "HRV_SDNN", "HRV_RMSSD", "HRV_pNN50"]])
```

```python
import numpy as np

rng = np.random.default_rng(0)
ecg_art = ecg.copy()

for t in rng.choice(len(ecg) - 15, 8, replace=False):
    ecg_art[t:t + 15] += rng.choice([-1, 1]) * 2.0

signals_a, info_a = nk.ecg_process(ecg_art, sampling_rate=300)
nk.ecg_plot(signals_a, info_a)
fig = plt.gcf(); fig.set_size_inches(15, 8); fig.tight_layout()
plt.show()

hrv_a = nk.hrv_time(info_a["ECG_R_Peaks"], sampling_rate=300)
print(hrv_a[["HRV_MeanNN", "HRV_SDNN", "HRV_RMSSD", "HRV_pNN50"]])
```
